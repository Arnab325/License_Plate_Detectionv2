```python
from ultralytics import YOLO

# Load pretrained YOLOv8m model
model = YOLO("yolov8m.pt")

# Train the model
results = model.train(
    data="data_custom.yaml",
    epochs=30,
    batch=8,
    imgsz=640,
    workers=8,

    # Augmentation
    hsv_h=0.05,
    hsv_s=0.9,
    hsv_v=0.7,
    translate=0.1,
    scale=0.5,
    flipud=0.0,
    fliplr=0.5,
    mosaic=0.0,
    mixup=0.0,

    # Warmup
    warmup_epochs=3.0,

    # Loss weights
    box=10.0,
    cls=0.3,
    dfl=0.99,

    # Validation and saving
    val=True,
    save=True,
    seed=0
)

print("Training completed successfully!")
```
